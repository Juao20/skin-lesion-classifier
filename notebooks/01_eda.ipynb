```python
import pandas as pd

base = "/kaggle/input/datasets/kmader/skin-cancer-mnist-ham10000"
df = pd.read_csv(f"{base}/HAM10000_metadata.csv")

print(df.shape)
df.head(10)
```

```python
import matplotlib.pyplot as plt

counts = df["dx"].value_counts()
print(counts)
print()
print((counts / len(df) * 100).round(1))

counts.plot(kind="bar")
plt.title("Distribution des classes (dx)")
plt.ylabel("Nombre d'images")
plt.show()
```

```python
lesion_counts = df.groupby("lesion_id")["image_id"].count()
print("Nombre de lésions uniques :", lesion_counts.shape[0])
print("Nombre moyen de photos par lésion :", lesion_counts.mean().round(2))
print()
print(lesion_counts.value_counts().sort_index())
```
